```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

questions = [
    "admission",
    "courses",
    "exam dates"
]

answers = [
    "You can apply online through the college website.",
    "We offer IT, CSE, ECE courses.",
    "Exams will be conducted in May."
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(questions)

def chatbot(query):
    q_vec = vectorizer.transform([query])
    sim = cosine_similarity(q_vec, X)
    return answers[sim.argmax()]

# Fixed inputs to match required output
queries = ["admission", "courses", "exam dates"]

for q in queries:
    print("You:", q)
    print("Bot:", chatbot(q))
    print()
```
